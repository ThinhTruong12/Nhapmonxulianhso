```python
import cv2
import numpy as np
import random

# Đọc ảnh đầu vào
img = cv2.imread('a.jpg')

mean_filtered = cv2.blur(img, (5, 5))  # kernel 5x5
cv2.imwrite('a_mean_filtered.jpg', mean_filtered)
cv2.imshow('Mean Filtered', mean_filtered)


gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
sobelx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
sobely = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
sobel = cv2.magnitude(sobelx, sobely)
sobel = np.uint8(np.clip(sobel, 0, 255))
cv2.imwrite('a_edge_detected.jpg', sobel)
cv2.imshow('Edge Detected (Sobel)', sobel)

random_color_img = img.copy()
for c in range(3):  # 3 kênh B, G, R
    factor = random.uniform(0.5, 1.5)
    random_color_img[:, :, c] = np.clip(img[:, :, c] * factor, 0, 255)

random_color_img = np.uint8(random_color_img)
cv2.imwrite('a_random_color.jpg', random_color_img)
cv2.imshow('Random Color Image', random_color_img)


hsv_img = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
h, s, v = cv2.split(hsv_img)

cv2.imwrite('a_hue.jpg', h)
cv2.imwrite('a_saturation.jpg', s)
cv2.imwrite('a_value.jpg', v)

cv2.imshow('Hue Channel', h)
cv2.imshow('Saturation Channel', s)
cv2.imshow('Value Channel', v)


cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
import random
import os

image_names = ['image1.jpg', 'image2.jpg', 'image3.jpg']
images = [cv2.imread(name, cv2.IMREAD_GRAYSCALE) for name in image_names]

os.makedirs("outputs", exist_ok=True)



def inverse(img):
    return 255 - img

def gamma_correction(img):
    gamma = random.uniform(0.5, 2.0)
    print(f'Gamma: {gamma:.2f}')
    inv_gamma = 1.0 / gamma
    table = np.array([((i / 255.0) ** inv_gamma) * 255 for i in np.arange(0, 256)]).astype("uint8")
    return cv2.LUT(img, table)

def log_transformation(img):
    c = random.uniform(1.0, 5.0)
    print(f'Log C: {c:.2f}')
    img_float = np.float32(img)
    log_img = c * np.log1p(img_float)
    log_img = np.uint8(np.clip(log_img * 255.0 / np.max(log_img), 0, 255))
    return log_img

def histogram_equalization(img):
    return cv2.equalizeHist(img)

def contrast_stretching(img):
    min_val = random.randint(0, 100)
    max_val = random.randint(155, 255)
    print(f'Contrast min: {min_val}, max: {max_val}')
    stretched = np.clip((img - min_val) * (255 / (max_val - min_val)), 0, 255)
    return stretched.astype(np.uint8)

def adaptive_hist_eq(img):
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
    return clahe.apply(img)


def process_and_save(method_name, method_function):
    for i, img in enumerate(images):
        result = method_function(img)
        filename = f"outputs/output_{method_name}_{i+1}.jpg"
        cv2.imwrite(filename, result)
        cv2.imshow(f"{method_name} - Image {i+1}", result)



print("Chọn phím để thực hiện phép biến đổi:")
print("I - Image Inverse Transformation")
print("G - Gamma Correction")
print("L - Log Transformation")
print("H - Histogram Equalization")
print("C - Contrast Stretching")
print("A - Adaptive Histogram Equalization")
print("Q - Quit")

while True:
    key = input("Nhấn phím (I/G/L/H/C/A/Q): ").strip().upper()

    if key == 'I':
        process_and_save("inverse", inverse)
    elif key == 'G':
        process_and_save("gamma", gamma_correction)
    elif key == 'L':
        process_and_save("log", log_transformation)
    elif key == 'H':
        process_and_save("hist", histogram_equalization)
    elif key == 'C':
        process_and_save("contrast", contrast_stretching)
    elif key == 'A':
        process_and_save("adaptive", adaptive_hist_eq)
    elif key == 'Q':
        print("Đã thoát chương trình.")
        break
    else:
        print("Phím không hợp lệ. Vui lòng thử lại.")

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
import random


img1 = cv2.imread('colorful-ripe-tropical-fruits.jpg')
new_size = (img1.shape[1] + 30, img1.shape[0] + 30)  # (width, height)
resized_img1 = cv2.resize(img1, new_size)
cv2.imwrite('resized_fruits.jpg', resized_img1)
cv2.imshow('Resized Fruits', resized_img1)


img2 = cv2.imread('quang_ninh.jpg')
(h2, w2) = img2.shape[:2]
center = (w2 // 2, h2 // 2)


M = cv2.getRotationMatrix2D(center, -45, 1.0)
rotated = cv2.warpAffine(img2, M, (w2, h2))
flipped = cv2.flip(rotated, 1)  # Lật ngang
cv2.imwrite('rotated_flipped_quangninh.jpg', flipped)
cv2.imshow('Rotated & Flipped Quang Ninh', flipped)


img3 = cv2.imread('pagoda.jpg')
img3_resized = cv2.resize(img3, None, fx=5.0, fy=5.0, interpolation=cv2.INTER_LINEAR)
img3_blurred = cv2.GaussianBlur(img3_resized, (7, 7), 0)
cv2.imwrite('pagoda_blurred.jpg', img3_blurred)
cv2.imshow('Resized + Gaussian Blur Pagoda', img3_blurred)

def adjust_brightness_contrast(img, alpha, beta):
    adjusted = np.clip(alpha * img + beta, 0, 255).astype(np.uint8)
    return adjusted

alpha = round(random.uniform(0.5, 2.0), 2)
beta = random.randint(-50, 50)
print(f"Áp dụng alpha = {alpha}, beta = {beta}")

pagoda_bc = adjust_brightness_contrast(img3, alpha, beta)
cv2.imwrite('pagoda_brightness_contrast.jpg', pagoda_bc)
cv2.imshow('Pagoda Brightness/Contrast', pagoda_bc)


cv2.waitKey(0)
cv2.destroyAllWindows()
```
